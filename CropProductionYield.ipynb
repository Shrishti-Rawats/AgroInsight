### Handling Missing Values in the `Production` Column

- The `Production` column contains 4,993 missing values.
- Among these, 4,960 records have a valid `Area` value and `Yield = 0`.
- To investigate whether these missing production values could represent zero production, existing records with `Yield = 0` and non-missing `Production` were examined.
- There are 1,024 such records, and **all 1,024 have `Production = 0`**.
- This consistent relationship indicates that in this dataset, a `Yield` value of 0 corresponds to zero production.
- Therefore, the 4,960 missing `Production` values associated with `Yield = 0` can reasonably be replaced with `0`.
- The remaining missing Production value will be investigated separately.

```python
# Replace missing Production with 0 when Yield is 0
df.loc[
    df['Production'].isnull() & (df['Yield'] == 0),
    'Production'
] = 0
```

8.6 Global IQR Analysis

```python
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"{'='*50}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Outlier Count:", len(outliers))
    print(
        "Outlier Percentage:",
        round(len(outliers) / len(df) * 100, 2),
        "%"
    )
```

8.7 Create Global Outlier Summary

```python
outlier_summary = []

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_percentage = (
        outlier_count / len(df)
    ) * 100

    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outlier_count,
        'Outlier Percentage': outlier_percentage
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

8.8 Crop-wise Yield Outlier Detection

```python
# Crop-wise Yield Outlier Detection

Q1 = df.groupby('Crop')['Yield'].transform('quantile', 0.25)
Q3 = df.groupby('Crop')['Yield'].transform('quantile', 0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['Yield_Outlier'] = (
    (df['Yield'] < lower_bound) |
    (df['Yield'] > upper_bound)
)
```
